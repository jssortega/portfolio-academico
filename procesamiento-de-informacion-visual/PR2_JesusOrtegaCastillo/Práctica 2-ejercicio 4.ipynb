4.	**Escriba un programa en Python que cargue un vídeo**

```python
import cv2, numpy as np

#Carga el video que le pasemos como parÃ¡metro en cap
cap = cv2.VideoCapture("video.avi")
```

**y lo muestre en pantalla con vuestro nombre superpuesto en la esquina superior derecha**

*Código:*

```python
color=(0,0,255)
grueso=3
while (cap.isOpened()):
    ret , frame = cap.read()
    
    if not ret:
        break;
        
    dimensiones=frame.shape
    alto = dimensiones[0]
    ancho = dimensiones[1]

    cv2.putText(frame,"Jesus Ortega Castillo",(ancho-350,50),cv2.FONT_HERSHEY_SIMPLEX,1,color,grueso)
    cv2.imshow('frame', frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

*Explicación:*

Empezando en el bucle, mientras cap que tiene el video esté abierto se ejecuta el bucle, en cada iteración guarda en las variables ret y frame cada imagen que compone el video, justo después de eso compruebo que ret tiene alguna imagen, es decir, que el video no ha acabado, en caso de que termine al entrar al if corta el bucle ya que si no daría un error puesto que lo siguiente al obtener el tamaño de la imagen en frame.shape no tendría ninguna.
Después obtengo las dimensiones de cada imagen para posteriormente usar el ancho para colocar el texto a la derecha y con el método de cv2 putText pongo un texto en cada imagen. En este caso pongo mi nombre en las coordenadas que le paso como tercer parámetro.
Lo siguiente es un condicional que espera un milisegundo antes de continuar y si se pulsa la tecla 'q' entra en el cuerpo del condicional y corta el bucle, lo que hace que finalice el video.
Una vez sale del bucle cierra la ventana.
