3.	**Escriba un programa en Python que, usando la biblioteca OpenCV, que use tres variables ancho alto y nombre y cree una imagen RGB de X píxeles de ancho por Y píxeles de alto**

```python
import numpy as np
import cv2, sys


#Guardamos los parametros
ancho = 800
alto = 800
nombre = "imagenResultadoDibujos.png"

#Creando imagen de las mismas dimensiones que la original
imagen = np.zeros((alto, ancho, 3), np.uint8)

#Con esto la ponemos en blanco, tambien imagen[:] = (255, 255, 255)
imagen.fill(255)

```

**dibuje en ella al menos dos líneas, dos polilíneas, dos circunferencias y dos círculos.**

*Código:*

```python
cian=(255,255,0)
rojo=(0,0,255)
amarillo=(0,255,255)
verde=(0,255,0)
grueso = 10;

pts = np.array([[40,200],  [40,600], [360,600],[360,200],[40,200]],np.int32)
pts = pts.reshape((-1,1,2))
cv2.polylines(imagen, [pts], True, amarillo, grueso)

cv2.ellipse(imagen,(400,200),(100,50),0,0,360,verde,grueso)
cv2.ellipse(imagen,(400,600),(100,50),0,0,360,verde,grueso)

i=20
while(i<ancho):
    cv2.line(imagen, (i,0), (i,alto), cian, grueso)
    cv2.circle(imagen, (int(ancho/2), int(alto/2)), i, rojo, grueso)
    i+=100
```

*Explicación:*

Para cada figura declaro un color diferente que posteriormente pasaré por parámetro, como también declaro un ancho en pixel, el mismo para todas las figuras.
Comienzo dibujando un cuadrado con polilíneas en el lado izquierdo, para ello en el array de puntos paso las 4 esquinas siendo la primera y la última la misma para cerrar el cuadrado.
Posteriormente dibujo dos elipses simétricas, les paso el centro de cada elipse alineado en el eje x y simétricos en el eje y con el mismo tamaño ambas.
Para las lineas y círculos los hago con un bucle para dibujar varias. En cuanto a las lineas las he hecho verticales a lo ancho de la imagen por eso el iterador, la variable i, es la que está en la coordenada x que es la que va variando y las coordenadas y son fijas ya que los puntos conectan desde el pixel más alto(0) al último que es el valor de la altura(800). Para las circunferencias teniendo todas el centro en el centro de la imagen lo que va aumentando es el radio que lo define el iterador.

**y muestre el resultado y lo grabe todo en una imagen png de nombre que esté en la variable nombre.**

```python
#Mostramos la imagen
r = cv2.imshow(nombre, imagen)

#La guardamos con el nombre que almacenamos en el argumento y termina la ejecuciÃ³n del programa
cv2.imwrite(nombre, imagen)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
