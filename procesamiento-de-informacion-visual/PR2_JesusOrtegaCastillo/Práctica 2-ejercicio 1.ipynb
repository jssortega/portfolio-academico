1. **Escriba un programa en Python que use dos variables nombreA y nombreB. en nombreA guardaremos el nombre de la imagen que se va a cargar y en nombreB estará el nombre de la imagen resultante de la transformación que haremos. Calcular las dimensiones de la imagen (alto, ancho y bandas). Las transformaciones que haremos ser una versión rotada de las mismas.**

```python
import numpy as np
import cv2, sys

nombreA="imagen.png"
nombreB="imagenResultado.png"

#Leemos la imagen del primer argumento y la cargamos en img
img = cv2.imread(nombreA)

#Si la imagen no se ha podido cargar, terminamos la ejecucion
if ( img is None ):
    print(" Error al cargar imagen ")
    sys.exit()
    
#Obtenemos los valores de la imagen en el array dimensiones, se podrÃ­a usar img.size
dimensiones=img.shape

#Filas o alto
alto = dimensiones[0]

#Columnas o ancho
ancho = dimensiones[1]

#Bandas
bandas = dimensiones[2]
```

* **Realice una versión de la misma rotada 90º, en sentido horario. Dicha rotación se llevará a cabo creando una imagen nueva y copiando los pixeles desde la imagen original en las posiciones adecuadas**


*Código:*

```python
#Creando imagen Rotada, para ello intercambiaremos el orden de alto y ancho en la llamada (1er parÃ¡metro es alto y 2do ancho)
imgTransformada = np.zeros((ancho, alto, bandas), np.uint8)

#Iteramos por la imagen y almacenamos el valor BGR de cada pixel y lo ponemos en la posicion adecuada
for i in range(0, alto):
    for j in range(0, ancho):
        imgTransformada.itemset((i, j, 0), img.item(alto-j-1, i, 0))
        imgTransformada.itemset((i, j, 1), img.item(alto-j-1, i, 1))
        imgTransformada.itemset((i, j, 2), img.item(alto-j-1, i, 2))
        
```

*Explicación de las operaciones realizadas:*

Hago 2 bucles for anidados que recorran la imagen original, si los píxeles están colocados en forma de matriz voy recorriendo cada fila de izquierda a derecha, cuando llega al final de la fila, sería j = ancho, la i que correspondería a las columnas sumaría 1 y recorrería la fila de abajo de nuevo.
Lo que hago en cada iteración es aplicar a imagen transformada el pixel que corresponde para girar la imagen, para ello hay que aplicar los 3 valores RGB que tiene cada pixel de la imagen original.
Para que rote hacia la derecha lo que hay que conseguir es que la última fila de la imagen original sea la primera columna de la imagen transformada, la penúltima fila la segunda columna... Así hasta recorrer todas las filas por eso al valor x(que sería la i) de la imagen transformada tiene que cambiar en cada iteración. Para ayuda así se vería de una forma gráfica:
    
    Original         Transformada

(1,1) (1,2) (1,3) --> (3,1) (2,1) (1,1)

(2,1) (2,2) (2,3) --> (3,2) (2,2) (1,2)

(3,1) (3,2) (3,3) --> (3,3) (2,3) (1,3)

**Por ultimo, mostraremos la imagen en pantalla y grabaremos el resultado en un fichero llamado nombreB**

```python
#Mostramos la imagen
r = cv2.imshow("Imagen transformada",imgTransformada)

#La guardamos con el nombre que elegimos y termina la ejecucion del programa
cv2.imwrite(nombreB,imgTransformada)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
