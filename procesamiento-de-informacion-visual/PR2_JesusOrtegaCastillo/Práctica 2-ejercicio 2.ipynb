2. **Escriba un programa en Python que use dos variables nombreA y nombreB. en nombreA guardaremos el nombre de la imagen que se va a cargar y en nombreB estará el nombre de la imagen resultante de la transformación que haremos. Calcular las dimensiones de la imagen (alto, ancho y bandas). Las transformaciones que haremos será una versión reflejada de la misma.**

```python
import numpy as np
import cv2, sys

nombreA="imagen.png"
nombreB="imagenEspejo.png"

#Leemos la imagen del primer argumento y la cargamos en img
img = cv2.imread(nombreA)

#Si la imagen no se ha podido cargar, terminamos la ejecucion
if ( img is None ):
    print(" Error al cargar imagen ")
    sys.exit()

#Obtenemos los valores de la imagen en el array dimensiones, se podrÃ­a usar img.size
dimensiones=img.shape

#Filas o alto
alto = dimensiones[0]

#Columnas o ancho
ancho = dimensiones[1]

#Bandas
bandas = dimensiones[2]
```

* **Realice una versión de la misma reflejada en el eje vertica. Dicha versión reflejada se llevará a cabo creando una imagen nueva y copiando los pixeles desde la imagen orignal en las posiciones adecuadas**

*Código:*

```python

#Creando imagen reflejada de las mismas dimensiones que la original
imgTransformada = np.zeros((alto, ancho, bandas), np.uint8)

#Iteramos por la imagen y almacenamos el valor BGR de cada pixel y lo ponemos en la posicion adecuada
for i in range(0, alto):
    for j in range(0, int(ancho/2)):
        imgTransformada.itemset((i,j,0), img.item(i,j,0))
        imgTransformada.itemset((i,j,1), img.item(i,j,1))
        imgTransformada.itemset((i,j,2), img.item(i,j,2))

for i in range(0, alto):
    for j in range(int(ancho/2), ancho):
        imgTransformada.itemset((i,j,0), img.item(i,ancho-j-1,0))
        imgTransformada.itemset((i,j,1), img.item(i,ancho-j-1,1))
        imgTransformada.itemset((i,j,2), img.item(i,ancho-j-1,2))
#RELLENAR
```

*Explicación de las operaciones realizadas:*

La mitad imagen de la izquierda es igual que la original po lo que empiezo con un bucle anidado colocando los píxeles igual que la original solo que al iterar el ancho itero solo la primera mita.
En cuanto al hacer la foto en espejo la mitad de la derecha comienza en el eje x con los píxeles de derecha a izquierda, por lo la iteración vertical, la variable i, es la misma pero para el ancho los píxeles comenzando por la izquierda se van poniendo los píxeles de la derecha de la foto original. Esta vez empieza a iterar desde la mitad hasta el final.

**Por ultimo, mostraremos la imagen en pantalla y grabaremos el resultado en un fichero llamado nombreB**

```python
#Mostramos la imagen
r = cv2.imshow("Imagen transformada",imgTransformada)

#La guardamos con el nombre que elegimos y termina la ejecucion del programa
cv2.imwrite(nombreB,imgTransformada)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
